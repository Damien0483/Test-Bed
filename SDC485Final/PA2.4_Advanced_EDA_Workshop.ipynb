Exploratory Data Analysis on Wine Quality Dataset - 7/10/2026 - DAMHAR2570

```python
# Step 1.2 Load Dataset
# Import Libraries

import pandas as pd
from datetime import datetime
from IPython.display import display, Markdown
import matplotlib.pyplot as plt
import seaborn as sns

# Basic Information
todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "DAMHAR2570" 

# Load Wine dataset
data = pd.read_csv('C:/Users/ecpi/Downloads/WineQuality.csv')
print("Wine Quality dataset loaded successfully")

# Display first few rows
display(Markdown("### Step 1.2: Load the Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
display(data.head())
```

```python
# Step 1.4 Data Cleaning
display(Markdown("### Step 1.4: Data Cleaning"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
missing_values = data.isnull().sum()
print(f"Missing values in each column:\n{missing_values}")

```

```python
# Step 2.1: Univariate Analysis
display(Markdown("### Step 2.1: Univariate Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
sns.set(style="whitegrid")

# Histograms
print("\n--- Histograms of Numerical Features ---")
data.hist(bins=20, figsize=(20, 15), layout=(4, 3))
plt.suptitle('Histograms of All Numerical Features', y=1.02)
plt.show()

#Bar Charts
print("\n--- Distribution of Wine Quality ---")
plt.figure(figsize=(10, 6))
sns.countplot(x='quality', data=data, palette='viridis')
plt.title('Distribution of Wine Quality Scores')
plt.xlabel('Quality Score')
plt.ylabel('Count')
plt.show()
```

```python
# Step 2.2: Bivariate Analysis
display(Markdown("### Step 2.2: Bivariate Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
numerical_features = data.drop('quality', axis=1).columns

#Boxplots
print("\n--- Feature Distributions by Wine Quality ---")
for feature in numerical_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='quality', y=feature, data=data, palette='pastel')
    plt.title(f'{feature.replace("_", " ").title()} by Wine Quality')
    plt.xlabel('Quality Score')
    plt.ylabel(feature.replace("_", " ").title())
    plt.show()
```

```python
# Step 3: Advanced Visualiztions
display(Markdown("### Step 3.1: Correlation Heatmaps"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

#Heatmap of the correlation matrix
plt.figure(figsize=(12, 10))
corr_matrix = data.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5, fmt=".2f")
plt.title('Heatmap of Correlation Matrix')
plt.show()

```

```python
# Step 3: Advanced Visualiztions
display(Markdown("### Step 3.2: Pair Plots for Feature Relationships"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

features_to_plot = data.drop('quality', axis=1)
pair_plot = sns.pairplot(features_to_plot, diag_kind='kde')
pair_plot.fig.suptitle("Pair Plots of Wine Features", y=1.02)
plt.show()
```

```python
# Step 3: Advanced Visualiztions
display(Markdown("### Step 3.3: Scatter Plot for Alcohol vs Quality"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

plt.figure(figsize=(10, 6))
sns.regplot(x='alcohol', y='quality', data=data, scatter_kws={'alpha':0.2})
plt.title('Relationship Between Alcohol Content and Wine Quality')
plt.xlabel('Alcohol Content')
plt.ylabel('Quality Score')
plt.show()
```

```python
# Step 3: Advanced Visualiztions
display(Markdown("### Step 3.4: Bivariate vs Target Variable"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

quality_corr = data.corr()['quality'].drop('quality').sort_values(ascending=False)
print("Correlation of each feature withWine Quality:")
print(quality_corr)

plt.figure(figsize=(10, 8))
sns.barplot(
    x=quality_corr.values,
    y=quality_corr.index,
    hue=quality_corr.index,      # assign hue
    palette='coolwarm',
    legend=False                 # disable legend
)
plt.title('Correlation of Features with Wine Quality')
plt.xlabel('Correlation Coefficient')
plt.ylabel('Features')
plt.show()
```
