```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from IPython.display import display, Markdown
todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "DAMHAR2570"
# Load the dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
data = pd.read_csv(url)
# Display dataset information
display(Markdown("### Step 1.2: Import Necessary Libraries and Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
print(f"Dataset shape: {data.shape}")
print(f"Dataset columns: {data.columns.tolist()}")
display(data.head())
```

```python
display(Markdown("### Step 2.1: Define Scenario Parameters"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Define scenario parameters
tip_percentage_scenarios = [0.10, 0.15, 0.20, 0.25]
total_bill_scenarios = np.arange(10, 101, 10)
# Display scenario parameters
print("Tip Percentage Scenarios:", tip_percentage_scenarios)
print("Total Bill Scenarios:", total_bill_scenarios)
```

```python
display(Markdown("### Step 2.2: Perform Scenario Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Perform scenario analysis
results = []
for tip_percentage in tip_percentage_scenarios:
    for total_bill in total_bill_scenarios:
        total_tip = total_bill * tip_percentage
        results.append({
            "Tip_Percentage": tip_percentage,
            "Total_Bill": total_bill,
            "Total_Tip": total_tip
        })
# Convert results to DataFrame
results_df = pd.DataFrame(results)
# Display results
display(results_df.head())
```

```python
display(Markdown("### Step 3.1: Visualize Tip Percentage Impact"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Plot the impact of tip percentage on total tip amount
plt.figure(figsize=(12, 6))
sns.lineplot(data=results_df, x="Total_Bill", y="Total_Tip", hue="Tip_Percentage", marker="o")
plt.title("Impact of Tip Percentage on Total Tip Amount")
plt.xlabel("Total Bill Amount ($)")
plt.ylabel("Total Tip Amount ($)")
plt.legend(title="Tip Percentage")
plt.grid(True)
plt.show()
```

```python
display(Markdown("### Step 3.2: Visualize Combined Scenarios"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Pivot the results for heatmap
pivot_table = results_df.pivot(index="Tip_Percentage", columns="Total_Bill", values="Total_Tip")
# Plot the heatmap of combined scenarios
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_table, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Heatmap of Combined Scenarios (Total Tip Amount)")
plt.xlabel("Total Bill Amount ($)")
plt.ylabel("Tip Percentage")
plt.show()
```

```python
display(Markdown("### Step 3.3: Compare Scenario Analysis to Actual Data"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Calculate the actual tip percentages in the dataset
data['Actual_Tip_Percentage'] = data['tip'] / data['total_bill']
# Calculate the average actual tip percentage for each total bill amount
actual_avg_tips = data.groupby('total_bill')['Actual_Tip_Percentage'].mean().reset_index()
# Plot the scenario analysis results
plt.figure(figsize=(12, 6))
sns.lineplot(data=results_df, x="Total_Bill", y="Total_Tip", hue="Tip_Percentage", marker="o")
# Plot the actual data on top of the scenario analysis
plt.scatter(actual_avg_tips['total_bill'], actual_avg_tips['Actual_Tip_Percentage'] * actual_avg_tips['total_bill'], color='red', label="Actual Data")
plt.xlim(0, 60)
plt.title("Comparison of Scenario Analysis and Actual Data")
plt.xlabel("Total Bill Amount ($)")
plt.ylabel("Total Tip Amount ($)")
plt.legend(title="Scenario / Actual Data")
plt.grid(True)
plt.show()
```
