```python
import pandas as pd
from datetime import datetime
import numpy as np
from IPython.display import Markdown, display

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "DAMHAR2570"

display(Markdown("### Step 2.1: Load the Census Income Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

data = pd.read_csv("C:/Users/ecpi/Downloads/census_income_dataset.csv")
print("Data head:", data.head)
```

```python
display(Markdown("### Step 2.2: Initial Exploration of Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

print("Data Info:")
data.info()

print("\nDescriptive Statistics:")
print(data.describe(include="all"))
```

```python
display(Markdown("### Step 3.1: Data Cleaning"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Check for missing values
print("Missing values per column:")
print(data.isnull().sum())

#Adjust for missing values drop row with missing values
data = data.dropna()

#Remove duplicate rows
before_dups = data.shape[0]
data = data.drop_duplicates()
after_dups = data.shape[0]

print(f"\nRows before removing duplicates: {before_dups}")
print(f"Rows after removing duplicates: {after_dups}")
```

```python
display(Markdown("### Step 3.1: Data Cleaning"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

#Convert target variable income to binary
data['income'] = data['income'].apply(lambda x: 1 if x == '>50k' else 0)

#Separate features and target
X = data.drop('income', axis=1)
y = data['income']

# Identify categorical columns
categorical_cols = X.select_dtypes(include=['object']).columns

# One-hot encode categorical features
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

print("Feature matrix shape after encoding:", X.shape)
print("Target distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

display(Markdown("### Step 3.1: Data Cleaning"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
```

```python
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

display(Markdown("### Step 4.1: KNN and Predict"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Initialize and train KNN model
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Predict on test set
y_pred_knn = knn.predict(X_test)

print("Sample predictions:", y_pred_knn[:20])

```

```python
from sklearn.metrics import(
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

display(Markdown("### Step 5.1: Metrics and Confusion Matrix"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

acc = accuracy_score(y_test, y_pred_knn)
prec = precision_score(y_test, y_pred_knn)
rec = recall_score(y_test, y_pred_knn)
f1 = f1_score(y_test, y_pred_knn)

print(f"Accuracy: {acc:0.4f}")
print(f"Precision: {prec:0.4f}")
print(f"Recall: {rec:0.4f}")
print(f"F1 Score: {f1:0.4f}")

cm = confusion_matrix(y_test, y_pred_knn)
print("\nConfusionvMatrix:\n", cm)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix - initial KNN")
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

display(Markdown("### Step 6.1: Hyperparameter Tuning for KNN with GridSearchCV"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski']
}

grid_search_knn = GridSearchCV(
    KNeighborsClassifier(),
    param_grid,
    cv=5,
    scoring='accuracy',
    refit=True,
    verbose=2
)

grid_search_knn.fit(X_train, y_train)

print("Best Parameters:", grid_search_knn.best_params_)
print("Best Cross-Validation Score:", grid_search_knn.best_score_)
```

```python
display(Markdown("### Step 6.2: Evaluation of Tuned KNN Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

knn_tuned = grid_search_knn.best_estimator_
y_pred_knn_tuned = knn_tuned.predict(X_test)

acc_t = accuracy_score(y_test, y_pred_knn_tuned)
prec_t = precision_score(y_test, y_pred_knn_tuned)
rec_t = recall_score(y_test, y_pred_knn_tuned)
f1_t = f1_score(y_test, y_pred_knn_tuned)

print(f"Tuned Accuracy: {acc_t:0.4f}")
print(f"Tuned Precision: {prec_t:0.4f}")
print(f"Tuned Recall: {rec_t:0.4f}")
print(f"Tuned F1 Score: {f1_t:0.4f}")

cm_t = confusion_matrix(y_test, y_pred_knn_tuned)
print("\nConfusion Matrix (Tuned):\n", cm_t)

disp_t = ConfusionMatrixDisplay(confusion_matrix=cm_t)
disp_t.plot(cmap="Greens")
plt.title("Confusion Matrix - Tuned KNN")
plt.show()
```

```python
import seaborn as sns

display(Markdown("### Step 7.1: Confusion Matrix Heatmap for Tuned KNN"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

plt.figure(figsize=(6, 5))
sns.heatmap(cm_t, annot=True, fmt="d", cmap="Purples")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix Heatmap - Tuned KNN")
plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

display(Markdown("### Step 7.2: ROC Curve and AUC for Tuned KNN"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Get predicted probabilities for positive class
y_pred_proba = knn_tuned.predict_proba(X_test)[:, 1]

fpr, tpr, _= roc_curve(y_test, y_pred_proba)
auc_score = roc_auc_score(y_test, y_pred_proba)

print(f"AUC Score: {auc_score:0.2f}")

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {auc_score:0.2f})")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Tuned KNN on Census Income")
plt.legend(loc="lower right")
plt.show()
```
