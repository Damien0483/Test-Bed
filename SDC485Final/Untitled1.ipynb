```python
import pandas as pd
from datetime import datetime
from IPython.display import display, Markdown

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

from xgboost import XGBClassifier
```

```python
# Step 1: Load and Explore the Dataset
todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "DAMHAR2570"

display(Markdown("### Step 1: Load and Explore the Breast Cancer Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Load the Breasy Cancer dataset 
data = pd.read_csv("C:/Users/ecpi/Downloads/breastCancerDataset.csv")

# Show first few rows
display(data.head())

# Explore structure
print("\nDataFrame info:")
print(data.info())

print("\nDescriptive statistics:")
display(data.describe())
```

```python
display(Markdown("### Step 2.1: Handlling Missing Values and Duplicates"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

#Check missing values
missing_values = data.isnull().sum()
print("Missing values in each column:\n", missing_values)

numeric_cols = data.select_dtypes(include=["float64", "int64"]).columns
categorical_cols = data.select_dtypes(include=["object"]).columns

for col in numeric_cols:
    data[col].fillna(data[col].median())

for col in categorical_cols:
    data[col].fillna(data[col].mode()[0])

# Remove duplicate rows
before_dupes = data.shape[0]
data = data.drop_duplicates()
after_dupes = data.shape[0]

print(f"\nRows before removing duplicates=: {before_dupes}")
print(f"Rows after removing duplicates: {after_dupes}")

# Verify no remaining missing values
missing_values_after = data.isnull().sum()
print("\nMissing values after cleaning:\n", missing_values_after)
    
```

```python
display(Markdown("### Step 2.2: Standardizing Concavity1 and Concavity3"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Initialize scaler
scaler = StandardScaler()

# Seelct features to standardize
features_to_scale = ["concavity1", "concavity3"]

# Apply scaling
data[features_to_scale] = scaler.fit_transform(data[features_to_scale])

# Verify changes
display(data[features_to_scale].head())
```

```python
display(Markdown("### Step 2.3: Splitting the Data 70/30 Train Test Split"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Prepare features X and target y
data["Diagnosis_encoded"] = data["Diagnosis"].map({"M": 1, "B": 0})

X = data.drop(columns=["Diagnosis", "Diagnosis_encoded", "ID"])
y = data["Diagnosis_encoded"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Test set size: {X_test.shape[0]} samples")
```

```python
display(Markdown("### Step 3: Implementing SVM"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Initialize SVM with default parameters
svm_model = SVC()

# Fit Model
svm_model.fit(X_train, y_train)

# Predict on test data
y_pred_svm = svm_model.predict(X_test)
```

```python
display(Markdown("### Step 4: Evaluating the SVM Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Accuracy
svm_accuracy = accuracy_score(y_test, y_pred_svm)
print("SVN Accuracy Score:", svm_accuracy)

# Classification report
print("\nSVM Classification Report:")
print(classification_report(y_test, y_pred_svm))
```

```python
display(Markdown("### Step 5: SVM Hyperparameter Tuning with GridSearchCV"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define parameter grid
param_grid_svm = {
    "C": [0.1, 1, 10, 100],
    "kernel": ["linear", "rbf", "poly"]
}

# Initialize GridSearchCV
grid_search_svm = GridSearchCV(
    SVC(),
    param_grid_svm,
    cv=5,
    verbose=2,
    n_jobs=-1
)

# Fit grid search
grid_search_svm.fit(X_train, y_train)

print("Best SVM Parameters:", grid_search_svm.best_params_)
print("Best SVM CV Score:", grid_search_svm.best_score_)

# Refit model using best parameters
best_svm = grid_search_svm.best_estimator_

# Predict with tuned model
y_pred_svm_tuned = best_svm.predict(X_test)
```

```python
display(Markdown("### Step 6: Evaluating the Tuned SVM Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Accuracy
svm_tuned_accuracy = accuracy_score(y_test, y_pred_svm_tuned)
print("Tuned SVM Accuracy Score:", svm_tuned_accuracy)

# Classification report
print("\nTuned SVM Classification Report:")
print(classification_report(y_test, y_pred_svm_tuned))
```

```python
display(Markdown("### Step 7: Implementing XGBoost (XGBClassifier)"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Initialize XGBClassifier 
xgb_model = XGBClassifier(
    eval_metric="logloss",
    random_state=42
)

xgb_model.fit(X_train, y_train)

y_pred_xgb = xgb_model.predict(X_test)
```

```python
display(Markdown("### Step 8: Evaluating the XGBoost Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Accuracy
xgb_accuracy = accuracy_score(y_test, y_pred_xgb)
print("XGBoost Accuracy Score:", xgb_accuracy)

# Classification report
print("\nXGBoost Classification Report:")
print(classification_report(y_test, y_pred_xgb))
```
